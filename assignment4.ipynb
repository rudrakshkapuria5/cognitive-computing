```python
import pandas as pd
```

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python
roll_number = "1024170316"

last_two_digits = roll_number[-2:]

categories = ["billing", "account", "general"]

digit1 = int(last_two_digits[0])
digit2 = int(last_two_digits[1])

category1 = categories[digit1 % 3]
category2 = categories[digit2 % 3]

category1, category2
```

```python
personalized_entries = [
    {
        "question": "how do i update my registered mobile number",
        "answer": "Go to Account Settings and update your registered mobile number.",
        "keywords": "mobile number update account",
        "category": category1
    },
    {
        "question": "how can i change my email address",
        "answer": "Go to Account Settings and change your registered email address.",
        "keywords": "email change account",
        "category": category2
    }
]
```

```python
faq_entries = fixed_entries + personalized_entries

df = pd.DataFrame(faq_entries)

print(df)
```

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for i in range(len(df)):
        keywords = df.loc[i, "keywords"].lower().split()
        score = 0

        for word in query_words:
            if word in keywords:
                score = score + 1

        if score > 0:
            results.append([i, score, df.loc[i, "question"], df.loc[i, "answer"]])

    results.sort(key=lambda x: x[1], reverse=True)

    return results
```

```python
query = input("Enter your query: ")

results = score_query(query, df)

for result in results:
    print("Question:", result[2])
    print("Answer:", result[3])
    print("Confidence:", result[1])
    print()
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name][["question", "category"]]
```

```python
def score_query_with_tie(query, df):
    query_words = query.lower().split()
    results = []

    for i in range(len(df)):
        keywords = df.loc[i, "keywords"].lower().split()
        score = 0

        for word in query_words:
            if word in keywords:
                score = score + 1

        if score > 0:
            results.append([i, score, df.loc[i, "question"], df.loc[i, "answer"]])

    if len(results) == 0:
        print("No matching entries found.")
        return

    results.sort(key=lambda x: x[1], reverse=True)

    highest_score = results[0][1]

    print("Matching entries:")

    for result in results:
        if result[1] == highest_score:
            print("Question:", result[2])
            print("Answer:", result[3])
            print("Confidence:", result[1])
            print()
```

```python
score_query_with_tie("fee", df)
```
