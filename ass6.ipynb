```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
M = float(input("Enter M:"))

x=np.linspace(-10,10,100)
y1 = M*x**2
y2 = M*np.sin(x)
```

```python
print(x)
```

```python
print(y1, y2)
```

```python
plt.plot(x,y1, color = "blue", linestyle="-", label = "M*x^2")
plt.plot(x,y2, color = "red", linestyle="--", label = "M*sin(x)")

plt.legend()
plt.grid()
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.show()


```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {

        "Subject":   ["Cognitive", "ML", "CAO", "AI RIsk", "Software"],
        "Score":  [92, 100, 77, 88, 99]
}

df = pd.DataFrame(data)
print(df)
```

```python
sns.barplot(x="Subject", y="Score", data = df, palette="Set2")

for i in range(len(df)):
  plt.text(i, df["Score"][i]+1, df["Score"][i], ha="center")



plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.arange(50)

noise = np.random.normal(0, 1, 50)


plt.subplot(2, 2, 2)

plt.scatter(range(50), data + noise)

plt.title("Scatter Plot with Random Noise")
plt.xlabel("Index")
plt.ylabel("Value")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
sns.lineplot(x="month_number", y="total_profit", data=df, marker="o")

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid()
plt.show()
```

```python
sns.lineplot(x="month_number", y="facecream", data=df, label="Face Cream")
sns.lineplot(x="month_number", y="facewash", data=df, label="Face Wash")
sns.lineplot(x="month_number", y="toothpaste", data=df, label="Toothpaste")
sns.lineplot(x="month_number", y="bathingsoap", data=df, label="Bathing Soap")
sns.lineplot(x="month_number", y="shampoo", data=df, label="Shampoo")
sns.lineplot(x="month_number", y="moisturizer", data=df, label="Moisturizer")

plt.title("Product Sales Data")
plt.xlabel("Month Number")
plt.ylabel("Sales Units")
plt.grid()
plt.legend()
plt.show()
```

```python
df.plot(x="month_number", kind="bar", figsize=(12, 6))

plt.title("Company Sales Data")
plt.xlabel("Month Number")
plt.ylabel("Values")
plt.grid(axis="y")
plt.show()
```

```python
sales_columns = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer",
    "total_units",
    "total_profit"
]

for column in sales_columns:
    sns.barplot(x="month_number", y=column, data=df)
    plt.title(column)
    plt.xlabel("Month Number")
    plt.ylabel(column)
    plt.grid(axis="y")
    plt.show()
```
